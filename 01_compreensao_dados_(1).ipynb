## 4. Compreensão dos dados

### 4.1 Dicionário de dados

Descrição das colunas das duas tabelas. O Portal da Transparência do SNS não fornece descrições escritas das colunas, então os significados abaixo foram deduzidos dos nomes, dos tipos de dados e do sistema de triagem de Manchester. Os pontos marcados como "a confirmar" são verificados ao longo da análise.

**Tabela de ocupação hospitalar**

| Coluna | Tipo | Significado |
|---|---|---|
| Período | Data (AAAA-MM) | Mês de referência |
| Região | Texto | Região de saúde da instituição (5 valores: Norte, Centro, LVT, Alentejo e Algarve). LVT = Lisboa e Vale do Tejo |
| Instituição | Texto | Nome do hospital, centro hospitalar ou unidade local de saúde. Ex.: "Hospital de Cascais, PPP" |
| Localização Geográfica | Coordenadas | Latitude e longitude da instituição |
| Nº Dias de Internamento | Inteiro | Total de dias de internamento acumulado desde janeiro do ano (recomeça em janeiro) |
| Lotação Praticada | Inteiro | Número de camas em funcionamento (a confirmar com fonte oficial) |
| Taxa Anual de Ocupação em Internamento | Decimal | Percentagem de ocupação acumulada no ano. Pode ultrapassar 100%, e alguns valores extremos precisam de verificação |

**Tabela de urgência (triagem de Manchester)**

| Coluna | Tipo | Significado |
|---|---|---|
| Período, Região, Instituição, Localização Geográfica | — | Iguais à tabela de ocupação |
| Nº Atendimentos ... Vermelha | Inteiro | Atendimentos emergentes (nível mais grave) |
| Nº Atendimentos ... Laranja | Inteiro | Atendimentos muito urgentes |
| Nº Atendimentos ... Amarela | Inteiro | Atendimentos urgentes |
| Nº Atendimentos ... Verde | Inteiro | Atendimentos pouco urgentes |
| Nº Atendimentos ... Azul | Inteiro | Atendimentos não urgentes |
| Nº Atendimentos ... Branca | Inteiro | Pulseira fora do sistema de Manchester, atribuída por razões administrativas ou casos clínicos específicos, segundo fontes jornalísticas (a confirmar em fonte oficial). Acumulado desde janeiro |
| Nº Atendimentos sem Triagem | Inteiro | Atendimentos sem classificação de cor |
| Nº Atendimentos ... (todas as cores e sem triagem) | Inteiro | Atendimentos acumulados desde janeiro do ano |

Observação: os valores de atendimentos podem também ser acumulados ao longo do ano (a confirmar).

### 4.7 Verificação: os valores são acumulados?
Teste com uma instituição de série completa (ULS de Matosinhos), observando os meses de novembro de 2024 a março de 2025, para verificar se os valores recomeçam em janeiro.

```python
inst = 'Unidade Local de Saúde de Matosinhos, EPE'
```

#### Dias de internamento

```python
teste_ocup = ocup[(ocup['Instituição'] == inst) &
                  (ocup['Período'] >= '2024-11') &
                  (ocup['Período'] <= '2025-03')]

teste_ocup[['Período', 'Nº Dias de Internamento', 'Lotação Praticada',
            'Taxa Anual de Ocupação em Internamento']].sort_values('Período')
```

#### Atendimentos na urgência

```python
col_verm = 'Nº Atendimentos em Urgência SU Triagem Manchester -Vermelha'
col_amar = 'Nº Atendimentos em Urgência SU Triagem Manchester -Amarela'

teste_urg = urg[(urg['Instituição'] == inst) &
                (urg['Período'] >= '2024-11') &
                (urg['Período'] <= '2025-03')]

teste_urg[['Período', col_verm, col_amar]].sort_values('Período')
```

**Conclusões da seção 4.7**

- Os dias de internamento e os atendimentos da urgência são valores acumulados desde janeiro de cada ano: na ULS de Matosinhos, os valores sobem até dezembro e recomeçam em janeiro (ex.: dias de internamento de 116.899 em 2024-12 para 10.481 em 2025-01).
- A "Taxa Anual de Ocupação" corresponde ao acumulado do ano: em 2024, dias acumulados ÷ (camas × dias decorridos) reproduz o valor do portal.
- Para análises mensais, os valores serão desacumulados na limpeza (valor do mês = acumulado − acumulado do mês anterior, exceto em janeiro) e a taxa de ocupação mensal será recalculada.

#### 4.7.2 Quedas dentro do mesmo ano
Num valor acumulado, o número só pode subir ou ficar igual dentro do mesmo ano. Contagem de quantas vezes ele diminui.

```python
ocup['ano'] = ocup['Período'].str[:4]
ocup['dif'] = ocup.groupby(['Instituição', 'ano'])['Nº Dias de Internamento'].diff()
print("Quedas na ocupação:", (ocup['dif'] < 0).sum())
```

```python
urg['ano'] = urg['Período'].str[:4]
urg_ordenado = urg.sort_values(['Instituição', 'Período'])

# Definir as colunas de atendimento antes de utilizá-las no loop
colunas_atend = [col for col in urg.columns if col.startswith('Nº Atendimentos')]

for col in colunas_atend:
    dif = urg_ordenado.groupby(['Instituição', 'ano'])[col].diff()
    print(col[-20:], "→ quedas:", (dif < 0).sum())
```
